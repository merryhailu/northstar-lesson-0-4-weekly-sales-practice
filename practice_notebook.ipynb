Jupiter Notebook

Jupyter Notebook helps us write code, see results, and do analysis all in one place. We can run and modify the code easily, then immediately see the output and results we want. It is also useful for sharing projects and analysis with colleagues because everything — code, explanations, and results — stays together on one page.

```python
weekly_sales = {
"week" : ["week 1", "week 2", "week 3", "week 4", "week 5", "week 6"],
"sales" : [8900, 6000, 5500, 6200, 5800, 6100]
}

print(weekly_sales)
```

```python
for week, sales in zip(weekly_sales['week'], weekly_sales['sales']):
    print(week , ":", sales)
    
```

```python
sum(weekly_sales["sales"])
```

```python
len(weekly_sales['sales'])
```

```python
max(weekly_sales["sales"])
```

```python
min(weekly_sales["sales"])
```

```python
import pandas as pd 

df = pd.DataFrame(weekly_sales)
print(df)
```
